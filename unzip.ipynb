```python
from pathlib import Path
from zipfile import ZipFile

# Thu muc chua file zip (mac dinh: thu muc hien tai)
zip_dir = Path('.')

# Neu muon giai nen mot file cu the, sua ten file o day
# zip_files = [zip_dir / 'your_file.zip']

# Giai nen tat ca file .zip trong thu muc
zip_files = list(zip_dir.glob('*.zip'))

if not zip_files:
    print('Khong tim thay file .zip nao trong thu muc hien tai.')
else:
    for zip_path in zip_files:
        # Tao thu muc dich theo ten file zip (bo duoi .zip)
        output_dir = zip_path.with_suffix('')
        output_dir.mkdir(parents=True, exist_ok=True)

        with ZipFile(zip_path, 'r') as zf:
            zf.extractall(output_dir)

        print(f'Da giai nen: {zip_path.name} -> {output_dir}')

    print('Hoan tat giai nen.')
```
